# 03 — Evaluation Analysis & Failure Modes

This notebook analyzes the benchmark evaluation results across all three systems:
1. **Trivial Baseline** (Majority class + canned response + always auto-handle)
2. **Simple Baseline** (TF-IDF + Logistic Regression + BM25 + Keyword escalation)
3. **Production AI Agent** (Hybrid Intent Classifier + Hybrid RAG + Multi-Factor Escalation Engine)


```
import json
import pandas as pd
from pathlib import Path

results_path = Path("../evaluation/eval_results.json")
if results_path.exists():
    with open(results_path, "r", encoding="utf-8") as f:
        res = json.load(f)
    df_comp = pd.DataFrame(res["baselines_comparison"]).T
    display(df_comp[["intent_accuracy", "intent_macro_f1", "escalation_accuracy", "escalation_f1", "missed_escalation_risk_rate", "judge_overall_mean"]])
else:
    print("Run `python -m evaluation.run_eval` to generate benchmark evaluation results.")
```
